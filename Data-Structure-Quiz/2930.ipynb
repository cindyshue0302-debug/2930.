````python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
import random
import time

def run_quiz_with_widgets(questions_data, duration):
    total_questions = len(questions_data)
    score = 0
    user_answers_log = []
    start_time = time.time()
    current_question_index = 0

    # Shuffle questions
    shuffled_questions = questions_data.copy()
    random.shuffle(shuffled_questions)

    # --- Widget Elements ---
    # Output widget to display results/feedback or general content dynamically
    quiz_area_output = widgets.Output()

    # Display elements for the current question
    timer_label = widgets.Label()
    progress_label = widgets.Label()
    question_text_label = widgets.Label()
    options_radio = widgets.RadioButtons(layout=widgets.Layout(width='auto'))
    submit_button = widgets.Button(description="下一題", button_style='info')

    # --- Layout --- (All quiz elements will be in this container)
    quiz_container = widgets.VBox([
        timer_label,
        progress_label,
        widgets.HTML('<hr>'), # Separator
        question_text_label,
        options_radio,
        widgets.HTML('<hr>'), # Separator
        submit_button,
        quiz_area_output # This is where feedback or results for each question will go
    ])

    # Function to update the timer display (called on each question display/submit)
    def update_timer_display():
        elapsed_time = time.time() - start_time
        remaining_time = max(0, duration - int(elapsed_time))
        mins, secs = divmod(remaining_time, 60)
        timer_label.value = f"剩餘時間: {mins:02d}分 {secs:02d}秒"
        if remaining_time <= 0:
            submit_button.disabled = True
            with quiz_area_output:
                clear_output()
                display(HTML("<h2 style='color:red;'>時間到！測驗自動結束。</h2>"))
            show_results()

    # Function to display the current question
    def show_question():
        nonlocal current_question_index, score, start_time

        # Check for timeout before displaying a new question
        if time.time() - start_time >= duration:
            submit_button.disabled = True
            with quiz_area_output:
                clear_output()
                display(HTML("<h2 style='color:red;'>時間到！測驗自動結束。</h2>"))
            show_results()
            return

        if current_question_index >= total_questions:
            show_results()
            return

        q_data = shuffled_questions[current_question_index]
        question_full_text = q_data["question"]

        # Remove leading number from question text for display
        display_question_text = question_full_text.split('. ', 1)[1] if question_full_text[0].isdigit() and '. ' in question_full_text else question_full_text
        question_text_label.value = f"<b>Q{current_question_index + 1}.</b> {display_question_text}"

        # Prepare options for RadioButtons widget
        # We need to show "A. Option Text" but store the full original option string as value
        options_for_widgets = []
        for i, opt_text_full in enumerate(q_data["options"]):
            # Robustly remove the (A), (B) etc. prefix for display
            idx_closing_paren = opt_text_full.find(')')
            if idx_closing_paren != -1 and idx_closing_paren + 1 < len(opt_text_full) and opt_text_full[idx_closing_paren + 1] == ' ':
                display_option_content = opt_text_full[idx_closing_paren + 2:]
            elif idx_closing_paren != -1:
                display_option_content = opt_text_full[idx_closing_paren + 1:]
            else:
                display_option_content = opt_text_full
            options_for_widgets.append((f"{chr(65 + i)}. {display_option_content}", opt_text_full))

        options_radio.options = options_for_widgets
        options_radio.value = None # Clear selection for the new question

        progress_label.value = f"進度: {current_question_index + 1} / {total_questions}"
        submit_button.description = "下一題" if current_question_index < total_questions - 1 else "完成測驗"
        submit_button.button_style = 'info'
        submit_button.disabled = False
        update_timer_display() # Update timer on question change

    # Event handler for the submit button
    def on_submit_button_clicked(b):
        nonlocal current_question_index, score

        # Check for timeout before processing answer
        if time.time() - start_time >= duration:
            submit_button.disabled = True
            with quiz_area_output:
                clear_output()
                display(HTML("<h2 style='color:red;'>時間到！測驗自動結束。</h2>"))
            show_results()
            return

        selected_answer = options_radio.value
        # Retrieve q_data before incrementing current_question_index
        q_data = shuffled_questions[current_question_index]
        question_full_text = q_data["question"]
        correct_answer_str = q_data["correct_answer"]
        explanation = q_data["explanation"]

        if selected_answer is None:
            with quiz_area_output:
                clear_output()
                display(HTML("<p style='color:orange; font-weight:bold;'>請選擇一個答案！</p>"))
            return # Don't proceed if no answer is selected

        is_correct = (selected_answer == correct_answer_str)
        if is_correct:
            score += 1

        # Determine user_choice_raw (A, B, C...) for logging
        user_choice_raw = "未作答"
        if selected_answer is not None:
            for i, opt_text in enumerate(q_data["options"]):
                if opt_text == selected_answer:
                    user_choice_raw = chr(65 + i)
                    break

        user_answers_log.append({
            "question": question_full_text,
            "user_choice_raw": user_choice_raw,
            "user_answer": selected_answer,
            "correct_answer": correct_answer_str,
            "is_correct": is_correct,
            "explanation": explanation
        })

        current_question_index += 1
        with quiz_area_output:
            clear_output() # Clear previous feedback

        # If all questions answered, show results directly
        if current_question_index >= total_questions:
            show_results()
        else:
            show_question()

    # Function to display final results
    def show_results():
        nonlocal score, total_questions, user_answers_log

        total_time_taken = time.time() - start_time
        mins = int(total_time_taken // 60)
        secs = int(total_time_taken % 60)

        # Clear the output of the cell where the quiz was displayed
        clear_output(wait=True)

        # Display results in a new Output widget
        final_results_output = widgets.Output()
        with final_results_output:
            display(HTML("<h1>測驗結束！</h1>"))
            display(HTML(f"<h2>您的總分： <span style='color:blue;'>{score}</span> / {total_questions}</h2>"))
            display(HTML(f"<h2>作答時間： {mins} 分 {secs} 秒</h2>"))
            display(HTML("<h3>--- 測驗結果詳情 ---</h3>"))

            for idx, result in enumerate(user_answers_log):
                status_mark = f"<span style='color:green;'>✔ 正確</span>" if result["is_correct"] else f"<span style='color:red;'>✘ 錯誤</span>"

                display_user_choice_content = ""
                if result["user_choice_raw"] == "未作答":
                    display_user_choice_content = "未作答"
                else:
                    opt_text_for_display = result["user_answer"]
                    idx_closing_paren = opt_text_for_display.find(')')
                    if idx_closing_paren != -1 and idx_closing_paren + 1 < len(opt_text_for_display) and opt_text_for_display[idx_closing_paren + 1] == ' ':
                        display_user_choice_content = opt_text_for_display[idx_closing_paren + 2:]
                    elif idx_closing_paren != -1:
                        display_user_choice_content = opt_text_for_display[idx_closing_paren + 1:]
                    else:
                        display_user_choice_content = opt_text_for_display
                    display_user_choice_content = f"{result['user_choice_raw']}. {display_user_choice_content}"

                display(HTML(f"<h4>第 {idx+1} 題: {result['question']}</h4>"))
                display(HTML(f"<p>您的作答： {display_user_choice_content} ({status_mark})</p>"))
                display(HTML(f"<p>正確答案： {result['correct_answer']}</p>"))
                display(HTML(f"<p>詳細解析： {result['explanation']}</p>"))
                display(HTML("<hr>"))
        display(final_results_output)

    # Attach the event handler to the button
    submit_button.on_click(on_submit_button_clicked)

    # Display the main quiz container
    display(quiz_container)

    # Start the quiz by showing the first question
    show_question()
````

````python
import random
import time
from IPython.display import display, HTML, clear_output
import sys

def run_quiz(questions, duration):
    total_questions = len(questions)
    score = 0
    user_answers_log = []
    start_time = time.time()

    # 隨機打亂題目順序
    random.shuffle(questions)

    # 初次清除，顯示測驗開始訊息
    clear_output(wait=True)
    print(f"測驗開始！您有 {duration // 60} 分鐘 {duration % 60} 秒完成 {total_questions} 題。")
    time.sleep(2)

    for i, q_data in enumerate(questions):
        elapsed_time = time.time() - start_time
        remaining_time = max(0, duration - int(elapsed_time))

        if remaining_time == 0:
            print("時間到！測驗自動結束。")
            break

        # 顯示當前問題的所有信息，不清空輸出，確保用戶可以看到輸入提示
        print(f"\n目前題數進度：第 {i+1} 題／共 {total_questions} 題") # Added newline for separation
        print(f"剩餘時間：{remaining_time // 60} 分 {remaining_time % 60} 秒")
        print("=" * 30)

        question_text = q_data["question"]
        # 移除問題開頭的題號 (例如 '1. ')
        display_question_text = question_text.split('. ', 1)[1] if question_text[0].isdigit() and '. ' in question_text else question_text

        original_options = q_data["options"]
        correct_answer_str = q_data["correct_answer"]
        explanation = q_data["explanation"]

        # 選項不需隨機排列
        # shuffled_options = random.sample(original_options, len(original_options))
        option_map = {chr(65 + idx): opt for idx, opt in enumerate(original_options)}

        print(display_question_text)
        for key, opt_text in option_map.items():
            # Robustly remove the (A), (B) etc. prefix, handling cases with or without a space after ')'
            idx_closing_paren = opt_text.find(')')
            if idx_closing_paren != -1 and idx_closing_paren + 1 < len(opt_text) and opt_text[idx_closing_paren + 1] == ' ':
                display_option_content = opt_text[idx_closing_paren + 2:]
            elif idx_closing_paren != -1:
                display_option_content = opt_text[idx_closing_paren + 1:]
            else:
                display_option_content = opt_text
            print(f"{key}. {display_option_content}")

        sys.stdout.flush() # Flush output to ensure everything is displayed before input

        user_input = ''
        valid_choices = [chr(65 + j) for j in range(len(original_options))] # 使用 original_options 的長度
        while user_input not in valid_choices:
            user_input = input("請輸入您的答案 (A, B, C...): ").upper()
            if user_input not in valid_choices:
                print(f"無效的輸入，請輸入 {', '.join(valid_choices)} 中的一個。")
                sys.stdout.flush() # Flush invalid input message as well

            # 再次檢查時間，避免用戶輸入太久導致超時後仍然讓其作答
            elapsed_time = time.time() - start_time
            remaining_time = max(0, duration - int(elapsed_time))
            if remaining_time == 0:
                print("時間到！測驗自動結束。")
                break

        if remaining_time == 0: # 如果是因為時間結束而跳出循環
            user_selected_option = "未作答 (時間到)"
            is_correct = False
        else:
            user_selected_option = option_map[user_input]
            is_correct = (user_selected_option == correct_answer_str)
            if is_correct:
                score += 1

        user_answers_log.append({
            "question": question_text,
            "user_choice_raw": user_input, # 儲存使用者輸入的字母
            "user_answer": user_selected_option,
            "correct_answer": correct_answer_str,
            "is_correct": is_correct,
            "explanation": explanation
        })

        # Removed clear_output from here to prevent hanging
        print("\n" + "=" * 30 + "\n") # Add a separator between questions

    end_time = time.time()
    total_time_taken = end_time - start_time

    clear_output(wait=True)
    print("=" * 30)
    print("測驗結束！")
    print("=" * 30)
    print(f"您的總分： {score} / {total_questions}")
    print(f"作答時間： {int(total_time_taken // 60)} 分 {int(total_time_taken % 60)} 秒")
    print("\n--- 測驗結果詳情 ---")

    for idx, result in enumerate(user_answers_log):
        status_mark = "✔ 正確" if result["is_correct"] else "✘ 錯誤"

        display_user_choice = ""
        if result["user_choice_raw"] == "": # 用戶未作答
            display_user_choice = "未作答"
        else:
            # Get the option text without the (A) prefix, consistent with how options were shown
            # Apply the same robust parsing as when displaying options
            opt_text_for_display = result["user_answer"]
            idx_closing_paren = opt_text_for_display.find(')')
            if idx_closing_paren != -1 and idx_closing_paren + 1 < len(opt_text_for_display) and opt_text_for_display[idx_closing_paren + 1] == ' ':
                chosen_option_content = opt_text_for_display[idx_closing_paren + 2:]
            elif idx_closing_paren != -1:
                chosen_option_content = opt_text_for_display[idx_closing_paren + 1:]
            else:
                chosen_option_content = opt_text_for_display
            display_user_choice = f"{result['user_choice_raw']}. {chosen_option_content}"

        print(f"\n第 {idx+1} 題: {result['question']}")
        print(f"您的作答： {display_user_choice} ({status_mark})")

        # Keep correct_answer as is (with its original A, B, C prefix)
        print(f"正確答案： {result['correct_answer']}")

        print(f"詳細解析： {result['explanation']}")
        print("-" * 20)
````
